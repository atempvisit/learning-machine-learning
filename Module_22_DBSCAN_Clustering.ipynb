```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler

from sklearn.datasets import make_moons
from sklearn.cluster import DBSCAN, KMeans
```

```python
df = make_moons()

df
```

```python
X, y = make_moons(n_samples=300, noise=0.06, random_state=42)

plt.scatter(X[:, 0], X[:, 1], c=y)
plt.show()
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
plt.scatter(X[:, 0], X[:, 1], c=y)
plt.show()
```

```python
dbscan = DBSCAN(eps=0.3, min_samples=3)
dbscan.fit(X_scaled)

label_dbscan = dbscan.labels_
label_dbscan
```

```python
sns.scatterplot( x=X[:, 0], y=X[:, 1], hue=label_dbscan, palette= 'viridis')
```

```python
kmeans = KMeans(
    n_clusters=2,
    init='k-means++',
    max_iter=300,
    random_state=42
)

kmeans.fit(X_scaled)

label_kmeans = kmeans.labels_
label_kmeans
```

```python
sns.scatterplot( x=X[:, 0], y=X[:, 1], hue=label_kmeans, palette= 'viridis')
```
